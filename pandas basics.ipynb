```python
import pandas as pd

# Creating a DataFrame from a dictionary
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'London', 'Paris']
}

df = pd.DataFrame(data)
print(df)
```

```python
import pandas as pd

# Reading a dataset from a CSV file
# df = pd.read_csv('data.csv')

# Displaying the first 5 rows of the DataFrame
print(df.head())

# Getting basic information about the data types and memory
print(df.info())
```

```python
# Filtering rows where 'Age' is greater than 28
older_people = df[df['Age'] > 28]
print(older_people)
```

```python
# Filling missing/null values with a specific number (e.g., 0)
df = df.fillna(0)

# Or dropping any row with missing values altogether
df = df.dropna()
```

```python
import pandas as pd

# Assume we have a DataFrame with 'City' and 'Salary'
data = {'City': ['Delhi', 'Mumbai', 'Delhi', 'Mumbai'],
        'Salary': [50000, 60000, 55000, 65000]}
df = pd.DataFrame(data)

# Calculating the average salary for each city
average_salary = df.groupby('City')['Salary'].mean()
print(average_salary)
```
